```python
# กลุ่ม O(n^2) - เข้าใจง่าย เหมาะกับ n เล็กหรือข้อมูลเกือบเรียงแล้วdef bubble_sort(a):    a = a[:]    n = len(a)    for i in range(n - 1):        swapped = False        for j in range(n - 1 - i):          # ตัวใหญ่สุด "ลอย" ไปท้ายทุกรอบ            if a[j] > a[j+1]:                a[j], a[j+1] = a[j+1], a[j]                swapped = True        if not swapped:                     # ถ้าไม่มีการสลับ = เรียงแล้ว -> best case O(n)            break    return adef selection_sort(a):    a = a[:]    for i in range(len(a)):        m = i        for j in range(i + 1, len(a)):      # หาตัวน้อยสุดในส่วนที่ยังไม่เรียง            if a[j] < a[m]: m = j        a[i], a[m] = a[m], a[i]             # จำนวน swap น้อยสุด = O(n)    return adef insertion_sort(a):    a = a[:]    for i in range(1, len(a)):        key, j = a[i], i - 1        while j >= 0 and a[j] > key:        # เลื่อนตัวที่ใหญ่กว่าไปทางขวา            a[j+1] = a[j]; j -= 1        a[j+1] = key    return adata = [64, 25, 12, 22, 11, 90, 34]print("bubble   :", bubble_sort(data))print("selection:", selection_sort(data))print("insertion:", insertion_sort(data))
```

```python
# กลุ่ม O(n log n) แบบ divide and conquerdef merge_sort(a):    if len(a) <= 1:        return a    mid = len(a) // 2    left, right = merge_sort(a[:mid]), merge_sort(a[mid:])    return merge(left, right)def merge(l, r):    out, i, j = [], 0, 0    while i < len(l) and j < len(r):        if l[i] <= r[j]:                    # <= ทำให้ stable            out.append(l[i]); i += 1        else:            out.append(r[j]); j += 1    out.extend(l[i:]); out.extend(r[j:])    return outdef quick_sort(a, lo=0, hi=None):    # in-place, Lomuto partition, สุ่ม pivot เพื่อเลี่ยง worst case O(n^2)    import random    a = a[:] if hi is None else a    if hi is None:        hi = len(a) - 1    def part(lo, hi):        k = random.randint(lo, hi)        a[k], a[hi] = a[hi], a[k]        pivot, i = a[hi], lo        for j in range(lo, hi):            if a[j] < pivot:                a[i], a[j] = a[j], a[i]; i += 1        a[i], a[hi] = a[hi], a[i]        return i    def go(lo, hi):        while lo < hi:            p = part(lo, hi)            if p - lo < hi - p:             # recursion ฝั่งเล็กก่อน -> ลึกไม่เกิน O(log n)                go(lo, p - 1); lo = p + 1            else:                go(p + 1, hi); hi = p - 1    go(lo, hi)    return aprint("merge:", merge_sort(data))print("quick:", quick_sort(data))
```

```python
# กลุ่มที่ไม่ใช้การเปรียบเทียบ (non-comparison sort)def counting_sort(a):    # เหมาะกับจำนวนเต็มไม่ติดลบที่ช่วงค่า (k) ไม่กว้าง -> O(n + k)    if not a: return []    k = max(a)    count = [0] * (k + 1)    for x in a: count[x] += 1    for i in range(1, k + 1): count[i] += count[i-1]     # prefix sum = ตำแหน่งปลาย    out = [0] * len(a)    for x in reversed(a):                                 # ไล่ย้อนเพื่อรักษา stability        count[x] -= 1        out[count[x]] = x    return outdef radix_sort(a):    # เรียงทีละหลัก โดยใช้ counting sort เป็นตัวช่วย -> O(d * (n + 10))    if not a: return []    out = a[:]    exp = 1    while max(out) // exp > 0:        buckets = [[] for _ in range(10)]        for x in out:            buckets[(x // exp) % 10].append(x)        out = [x for b in buckets for x in b]        exp *= 10    return outprint("counting:", counting_sort([4, 2, 2, 8, 3, 3, 1]))print("radix   :", radix_sort([170, 45, 75, 90, 802, 24, 2, 66]))
```
