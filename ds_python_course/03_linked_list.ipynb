## 3.1 แนวคิดLinked list เก็บข้อมูลเป็น **node** ที่กระจายในหน่วยความจำ แต่ละ node เก็บ `data` และ `next` (pointer ไป node ถัดไป)**เทียบกับ array**| ประเด็น | Array / list | Linked list ||---|---|---|| เข้าถึงตำแหน่งที่ i | \(O(1)\) | \(O(n)\) เดินไล่ || แทรก/ลบที่หัว | \(O(n)\) | \(O(1)\) || หน่วยความจำ | ต่อเนื่อง, ไม่มี overhead pointer | มี overhead ต่อ node || ขยายขนาด | ต้อง resize | ต่อ node เพิ่มได้ทันที |**ชนิดที่พบบ่อย:** Singly, Doubly, Circular

```python
class Node:    # node ของ singly linked list    __slots__ = ('data', 'next')          # ประหยัดหน่วยความจำ    def __init__(self, data, nxt=None):        self.data = data        self.next = nxtclass SinglyLinkedList:    def __init__(self):        self.head = None        self.size = 0    # ---------- insert ----------    def push_front(self, data):             # O(1)        self.head = Node(data, self.head)        self.size += 1    def push_back(self, data):              # O(n) เพราะต้องเดินไปท้าย        new = Node(data)        if not self.head:            self.head = new        else:            cur = self.head            while cur.next:                cur = cur.next            cur.next = new        self.size += 1    def insert_at(self, index, data):       # O(n)        if index <= 0:            return self.push_front(data)        if index >= self.size:            return self.push_back(data)        cur = self.head        for _ in range(index - 1):            cur = cur.next        cur.next = Node(data, cur.next)        self.size += 1    # ---------- delete ----------    def remove(self, data):                 # O(n) ลบตัวแรกที่เจอ        prev, cur = None, self.head        while cur:            if cur.data == data:                if prev is None:                    self.head = cur.next                else:                    prev.next = cur.next                self.size -= 1                return True            prev, cur = cur, cur.next        return False    # ---------- search / traverse ----------    def find(self, data):                   # O(n) คืน index หรือ -1        cur, i = self.head, 0        while cur:            if cur.data == data:                return i            cur, i = cur.next, i + 1        return -1    def __iter__(self):        cur = self.head        while cur:            yield cur.data            cur = cur.next    def __len__(self):        return self.size    def __repr__(self):        return " -> ".join(map(str, self)) + " -> None"ll = SinglyLinkedList()for x in (30, 20, 10):    ll.push_front(x)ll.push_back(40)ll.insert_at(2, 15)print(ll, "| len =", len(ll))print("find(15) =", ll.find(15), "| remove(20) =", ll.remove(20))print(ll)
```

## 3.2 เทคนิค Two Pointers บน Linked List- **Slow / Fast pointer** (Floyd) – slow เดิน 1 ก้าว, fast เดิน 2 ก้าว  - หา **จุดกลาง** ของ list  - ตรวจ **cycle** (วงวน)  - หา node ที่ k จากท้าย

```python
def middle_node(head):    # slow อยู่กลางเมื่อ fast ถึงท้าย -> O(n) time, O(1) space    slow = fast = head    while fast and fast.next:        slow, fast = slow.next, fast.next.next    return slowdef has_cycle(head):    # Floyd's cycle detection: ถ้ามีวง slow กับ fast จะมาเจอกัน    slow = fast = head    while fast and fast.next:        slow, fast = slow.next, fast.next.next        if slow is fast:            return True    return Falsedef nth_from_end(head, k):    # ให้ fast นำหน้า k ก้าว แล้วเดินพร้อมกัน    fast = head    for _ in range(k):        if not fast:            return None        fast = fast.next    slow = head    while fast:        slow, fast = slow.next, fast.next    return slowdef reverse_list(head):    # กลับทิศทาง pointer ทีละ node -> O(n) time, O(1) space    prev, cur = None, head    while cur:        cur.next, prev, cur = prev, cur, cur.next    return prevlst = SinglyLinkedList()for x in range(1, 8):    lst.push_back(x)print("list      :", lst)print("middle    :", middle_node(lst.head).data)print("2 จากท้าย  :", nth_from_end(lst.head, 2).data)print("มี cycle?  :", has_cycle(lst.head))lst.head = reverse_list(lst.head)print("reversed  :", lst)
```

```python
# Doubly Linked List: มี pointer ทั้ง prev และ next -> ลบ node ที่ถืออยู่ได้ O(1)class DNode:    __slots__ = ('data', 'prev', 'next')    def __init__(self, data):        self.data, self.prev, self.next = data, None, Noneclass DoublyLinkedList:    def __init__(self):        self.head = self.tail = None        self.size = 0    def push_back(self, data):              # O(1) เพราะเก็บ tail ไว้        n = DNode(data)        if not self.tail:            self.head = self.tail = n        else:            n.prev, self.tail.next = self.tail, n            self.tail = n        self.size += 1        return n    def unlink(self, node):                 # O(1) ลบเมื่อรู้ตัว node        if node.prev: node.prev.next = node.next        else:         self.head = node.next        if node.next: node.next.prev = node.prev        else:         self.tail = node.prev        self.size -= 1    def forward(self):        cur, out = self.head, []        while cur:            out.append(cur.data); cur = cur.next        return out    def backward(self):        cur, out = self.tail, []        while cur:            out.append(cur.data); cur = cur.prev        return outd = DoublyLinkedList()nodes = [d.push_back(c) for c in "ABCDE"]print("ไปหน้า :", d.forward())print("ย้อนกลับ:", d.backward())d.unlink(nodes[2])                          # ลบ 'C' แบบ O(1)print("หลังลบ C:", d.forward())
```
