## 9.1 การแทนกราฟกราฟ \(G = (V, E)\) มีจุด (vertex) และเส้นเชื่อม (edge) อาจเป็น *directed/undirected*, *weighted/unweighted*| การแทน | หน่วยความจำ | ตรวจว่ามี edge (u,v) | ไล่เพื่อนบ้านของ u ||---|---|---|---|| Adjacency **matrix** | \(O(V^2)\) | \(O(1)\) | \(O(V)\) || Adjacency **list** | \(O(V+E)\) | \(O(\deg u)\) | \(O(\deg u)\) |กราฟจริงส่วนใหญ่ *sparse* → นิยมใช้ adjacency list

```python
from collections import defaultdict, dequeclass Graph:    def __init__(self, directed=False):        self.adj = defaultdict(list)        self.directed = directed    def add_edge(self, u, v, w=1):        self.adj[u].append((v, w))        if not self.directed:            self.adj[v].append((u, w))        else:            self.adj.setdefault(v, [])    def vertices(self):        return list(self.adj.keys())    def matrix(self):        # แปลงเป็น adjacency matrix เพื่อเปรียบเทียบสองรูปแบบ        vs = sorted(self.adj)        idx = {v: i for i, v in enumerate(vs)}        M = [[0] * len(vs) for _ in vs]        for u in vs:            for v, w in self.adj[u]:                M[idx[u]][idx[v]] = w        return vs, Mg = Graph()for u, v in [('A','B'), ('A','C'), ('B','D'), ('C','D'), ('D','E')]:    g.add_edge(u, v)print("adjacency list:")for u in sorted(g.adj):    print(f"  {u}: {[v for v, _ in g.adj[u]]}")vs, M = g.matrix()print("adjacency matrix:", vs)for row in M: print("  ", row)
```

## 9.2 BFS และ DFS| | BFS | DFS ||---|---|---|| โครงสร้างช่วย | Queue | Stack / recursion || ลำดับสำรวจ | ทีละชั้นจากจุดเริ่ม | ลงลึกให้สุดก่อนถอยกลับ || Complexity | \(O(V+E)\) | \(O(V+E)\) || ใช้ทำ | เส้นทางสั้นสุดบนกราฟไร้น้ำหนัก, หาระดับชั้น | ตรวจ cycle, topological sort, หา connected component |

```python
from collections import dequedef bfs(g, start):    visited, order, q = {start}, [], deque([start])    while q:        u = q.popleft()        order.append(u)        for v, _ in g.adj[u]:            if v not in visited:                visited.add(v)          # mark ตอน enqueue กัน node เข้าคิวซ้ำ                q.append(v)    return orderdef bfs_shortest_path(g, start, goal):    # กราฟไร้น้ำหนัก: BFS ให้เส้นทางที่สั้นสุด (นับจำนวน edge)    prev, q = {start: None}, deque([start])    while q:        u = q.popleft()        if u == goal:            path = []            while u is not None:                path.append(u); u = prev[u]            return path[::-1]        for v, _ in g.adj[u]:            if v not in prev:                prev[v] = u; q.append(v)    return Nonedef dfs_recursive(g, u, visited=None, order=None):    visited = set() if visited is None else visited    order = [] if order is None else order    visited.add(u); order.append(u)    for v, _ in g.adj[u]:        if v not in visited:            dfs_recursive(g, v, visited, order)    return orderdef dfs_iterative(g, start):    visited, order, st = set(), [], [start]    while st:        u = st.pop()        if u in visited: continue        visited.add(u); order.append(u)        for v, _ in reversed(g.adj[u]):            if v not in visited: st.append(v)    return orderprint("BFS      :", bfs(g, 'A'))print("DFS (rec):", dfs_recursive(g, 'A'))print("DFS (it) :", dfs_iterative(g, 'A'))print("path A->E:", bfs_shortest_path(g, 'A', 'E'))
```
